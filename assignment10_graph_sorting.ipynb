## GRAPH 05

**문제**: 모든 정점 사이에 간선이 존재하는 그래프가 있다. 인접 행렬 표현과 인접 리스트 표현 중 어느 것을 사용할 것인가?

**답**: **인접 행렬**을 사용한다.

**이유**

모든 정점 사이에 간선이 존재하는 그래프는 **완전 그래프**에 가깝다. 즉, 간선의 수가 매우 많다. 정점 수를 \(V\), 간선 수를 \(E\)라고 하면, 인접 행렬은 항상 \(V 	imes V\) 크기의 공간을 사용한다. 간선이 매우 많은 경우에는 어차피 많은 연결 정보를 저장해야 하므로 인접 행렬의 공간 낭비가 상대적으로 작다.

또한 인접 행렬은 두 정점 \(u, v\) 사이에 간선이 있는지 확인할 때 `matrix[u][v]`만 보면 되므로 \(O(1)\) 시간에 확인할 수 있다. 따라서 간선 존재 여부를 빠르게 확인해야 하는 조밀 그래프에서는 인접 행렬이 적합하다.


## GRAPH 06

**문제**: 정점들 사이에 아주 드물게 간선이 존재하는 희소 그래프가 있다. 인접 행렬 표현과 인접 리스트 표현 중 어느 것을 사용할 것인가?

**답**: **인접 리스트**를 사용한다.

**이유**

희소 그래프는 정점 수에 비해 간선 수가 매우 적다. 인접 행렬을 사용하면 실제 간선이 거의 없어도 \(V^2\)개의 칸을 모두 만들어야 하므로 공간 낭비가 크다.

반면 인접 리스트는 실제 존재하는 간선만 저장하므로 공간 복잡도가 대략 \(O(V+E)\)이다. 따라서 간선 수 \(E\)가 적은 희소 그래프에서는 인접 리스트가 훨씬 효율적이다.


## GRAPH 08

**문제**: 정점 1을 시작으로 DFS(1)을 수행할 때, `v`가 정점 1인 상태에서 `(1) DFS(x)`는 몇 번 호출되는가?

첨부 PDF에서는 08번에 필요한 그래프 그림이 보이지 않는다. 따라서 정확한 숫자는 원본 그래프의 정점 1 인접 정점 구조가 있어야 계산할 수 있다.

### 풀이 원리

DFS 코드는 다음 구조이다.

```text
DFS(v):
    v.visited <- YES
    for each x ∈ v.adj
        if x.visited = NO
            DFS(x)
```

DFS(1)을 수행하면 정점 1을 먼저 방문 처리한다.
이후 정점 1의 인접 정점 중 아직 방문하지 않은 첫 번째 정점에 대해 DFS(x)가 호출된다.
이 재귀 호출 과정에서 그래프의 나머지 정점들이 모두 방문될 수 있으므로,
DFS(1)로 다시 돌아왔을 때 정점 1의 다른 인접 정점들은 이미 방문된 상태이다.

따라서 모함수 DFS(v)에서 v = 1인 경우에 실행되는 DFS(x)는 1번이다.

### STACK을 사용한 위상 정렬 수도 코드

```text
TopologicalSort_UsingStack(G):
    for each vertex v in G:
        indegree[v] <- 0

    for each vertex u in G:
        for each vertex v in Adj[u]:
            indegree[v] <- indegree[v] + 1

    stack <- empty stack

    for each vertex v in G:
        if indegree[v] = 0:
            push(stack, v)

    result <- empty list

    while stack is not empty:
        u <- pop(stack)
        append u to result

        for each vertex v in Adj[u]:
            indegree[v] <- indegree[v] - 1
            if indegree[v] = 0:
                push(stack, v)

    if length(result) < number of vertices:
        print("cycle exists")
    else:
        print(result)
```

### 가능한 위상 정렬 순서의 개수

위 간선 조건을 모두 만족하는 위상 정렬 순서를 세면 총 **63가지**이다.

**정답: 63가지**

## GRAPH 18

**문제**: 음의 가중치가 있는 그래프에서 다익스트라 알고리즘이 작동하지 않는 예를 3개의 정점으로 보이시오.

다음 방향 그래프를 예로 들 수 있다.

```text
A → B : 1
A → C : 2
C → B : -3
```

시작 정점을 A라고 하자.

다익스트라 알고리즘은 처음에 A에서 직접 갈 수 있는 거리로 다음을 저장한다.

```text
d(B) = 1
d(C) = 2
```

이때 가장 작은 거리는 B의 1이므로, B를 먼저 확정한다. 그런데 실제로는 다음 경로가 더 짧다.

```text
A → C → B = 2 + (-3) = -1
```

즉, B까지의 실제 최단 거리는 1이 아니라 -1이다. 하지만 다익스트라 알고리즘은 이미 B를 확정해 버렸기 때문에 나중에 더 짧은 경로가 발견되어도 제대로 반영하지 못한다.

따라서 **음의 가중치가 있는 그래프에서는 다익스트라 알고리즘이 올바르게 작동하지 않을 수 있다.**

````python
# 검산용 코드: MST, 위상 정렬 개수, 다익스트라 최단 거리 확인
from itertools import permutations

# GRAPH 12 MST - Kruskal
vertices = list("ABCDEFGH")
edges = [
    ("A","B",8), ("B","C",10), ("B","E",15), ("A","E",9),
    ("A","H",11), ("H","E",3), ("E","C",1), ("C","D",2),
    ("D","F",5), ("E","F",12), ("H","F",8), ("H","G",8),
    ("G","F",7), ("G","D",4)
]

parent = {v: v for v in vertices}

def find(x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]
    return x

def union(a, b):
    ra, rb = find(a), find(b)
    if ra == rb:
        return False
    parent[rb] = ra
    return True

mst = []
for u, v, w in sorted(edges, key=lambda x: x[2]):
    if union(u, v):
        mst.append((u, v, w))

print("MST:", mst)
print("MST total weight:", sum(w for _, _, w in mst))

# GRAPH 16 topological sort count
nodes = list("abcdefgh")
directed_edges = [
    ("a","d"), ("a","b"), ("b","c"), ("d","e"), ("d","f"),
    ("f","e"), ("e","g"), ("f","g"), ("f","h")
]

count = 0
for order in permutations(nodes):
    pos = {v: i for i, v in enumerate(order)}
    if all(pos[u] < pos[v] for u, v in directed_edges):
        count += 1
print("Topological sort count:", count)

# GRAPH 17 Dijkstra
INF = 10**9
graph = {
    "Q": [("A",10), ("C",7), ("E",4)],
    "A": [("B",8)],
    "B": [("D",12)],
    "C": [("A",2), ("B",8), ("F",12)],
    "D": [("G",4)],
    "E": [("C",2), ("F",8), ("G",20)],
    "F": [("D",7), ("G",3)],
    "G": []
}

dist = {v: INF for v in graph}
dist["Q"] = 0
closed = []

while len(closed) < len(graph):
    u = min([v for v in graph if v not in closed], key=lambda v: dist[v])
    closed.append(u)
    for v, w in graph[u]:
        if v not in closed and dist[u] + w < dist[v]:
            dist[v] = dist[u] + w

print("Dijkstra order:", closed)
print("Distances:", dist)
````
